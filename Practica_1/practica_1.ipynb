## Introducción

En el desarrollo de proyectos de robótica es necesario utilizar diferentes herramientas de software para **crear archivos, automatizar tareas, documentar procedimientos y compartir el trabajo**. Aunque estas herramientas tienen funciones diferentes, su combinación permite establecer flujos de trabajo reproducibles y facilita la colaboración.

- <img src="imagenes/Ubuntu-logo.png" height=60/> **Ubuntu** es un sistema operativo basado en Linux ampliamente utilizado en robótica. El manejo de la terminal permite ejecutar programas, administrar archivos y automatizar tareas sin depender exclusivamente de interfaces gráficas.

- **Bash (Bourne Again SHell)** es un intérprete de comandos utilizado habitualmente en sistemas Linux. Además de permitir ejecutar comandos de manera interactiva, permite crear **scripts de Bash**, es decir, archivos de texto que contienen una secuencia de comandos que puede ejecutarse de manera automatizada.

- <img src="imagenes/Jupyter-logo.png" height=60/> **Jupyter** proporciona un entorno interactivo basado en notebooks. Un notebook está formado por celdas que pueden contener texto, ecuaciones, código y resultados. En esta práctica se utilizará para documentar el trabajo y ejecutar comandos de Bash desde el mismo documento.

- **Markdown** es un lenguaje de marcado ligero que permite dar formato al texto mediante una sintaxis sencilla. Jupyter utiliza Markdown para crear títulos, listas, tablas, enlaces y ecuaciones dentro de las celdas de texto.

- <img src="imagenes/Git-logo.png" height=60/> **Git** es un sistema de control de versiones que permite registrar cambios en los archivos de un proyecto, recuperar versiones anteriores y trabajar mediante ramas.

- <img src="imagenes/Github-logo.png" height=60/> **GitHub** es una plataforma que permite alojar repositorios Git en línea y facilita el trabajo colaborativo mediante repositorios, ramas, colaboradores y herramientas de revisión.

En robótica, estas herramientas pueden utilizarse conjuntamente. Por ejemplo, un alumno puede crear un script de Bash para preparar una carpeta de trabajo, documentar el procedimiento en Jupyter y finalmente almacenar el notebook y los scripts en un repositorio de GitHub.


## Cuestionario previo

### Responder de forma breve las siguientes preguntas:

- En Linux, ¿qué es una **shell** y qué relación existe entre `sh` y `bash`?
> Un shell es un tipo de programa informático llamado intérprete de comandos que permite a los usuarios de Linux y Unix controlar sus sistemas operativos con interfaces de línea de comandos.  Hay dos tipos principales de shells, "shell" (sh) y "bash". Ambos se ejecutan en el sistema operativo Unix. Shell es un nombre genérico para los scripts de cualquier shell. Como su nombre indica, bash (Bourne Again Shell) es un shell mejorado y ampliado.  Bash utiliza la up-gradation para una mayor funcionalidad, admite controles de trabajos, es más fácil de usar que sh y admite el historial de comandos.  En muchos sistemas Linux, sh es simplemente un enlace simbólico que apunta a bash (o a una versión ligera de este en modo de compatibilidad).



- ¿Qué es un **script de Bash** y cuál es la diferencia entre ejecutar un comando directamente en la terminal y ejecutarlo mediante un script?

> Los Scripts Bash o Bash Scripts son simples archivos de texto que contienen listas de comandos que deben ser ejecutados en secuencia por el shell. Se escriben en archivos de texto plano e invocarlos es tan sencillo como referenciarlos en un comando de terminal. Al ejecutar un comando directamente en la terminal es procesado de forma interactiva e inmediata en la sesión actual. Ejecutar un script procesa múltiples comandos de forma no interactiva e integrada en una sub-shell (subproceso), lo que permite automatizar tareas complejas, reutilizar código y manejar estructuras lógicas (bucles, condiciones).


- ¿Qué es **Jupyter** y qué tipo de información puede contener una celda de un notebook?

> Un Jupyter Notebook es una aplicación web de código abierto que permite a los científicos de datos crear y compartir documentos que incluyen código en vivo, ecuaciones y otros recursos multimedia. Jupyter utiliza este .ipynbformato de archivo para almacenar los archivos de los cuadernos en tu ordenador (donde se guardan los datos y los resultados del cuaderno para su almacenamiento a largo plazo). El formato .ipynb es interpretado y modificado por la biblioteca de software nbformat.


- ¿Qué es **Markdown** y para qué se utiliza dentro de Jupyter?

> Markdown es un lenguaje de marcado que sirve para especificar el formato que debe tener un texto, en lo que respecta a elementos como títulos, negritas, listas y cosas similares. Markdown se usan unos códigos especiales, como saltos de línea, asteriscos o guiones, lo que lo hace mucho más ligero y sencillo de usar. En jupyter se utiliza para documentar el código introduciendo títulos, explicaciones, listas, enlaces, imágenes y fórmulas matemáticas (Latex), convirtiendo el notebook en un reporte reproducible y legible. 

- ¿Cuál es la diferencia entre **Git** y **GitHub**?

> Es el sistema de control de versiones distribuido que se instala localmente para rastrear los cambios en los archivos de un proyecto a lo largo del tiempo. Es una plataforma en la nube que aloja repositorios de Git, proporcionando una interfaz gráfica, herramientas de colaboración en equipo (Pull Requests, Issues, Actions) y respaldo remoto.

2. *Código del Script de Bash*

```bash
# Mensaje de inicio
echo "Bienvenido. Iniciando script de automatización para la Práctica 1"

# Carpeta de usuario
cd $HOME

# Verificar si existe la carpeta Practica1; si no existe, se crea
if [ ! -d "Practica1" ]; then
    echo "La carpeta Practica1 no existe. Creándola..."
    mkdir Practica1
else
    echo "La carpeta Practica1 ya existe."
fi

# Dentro de Practica1, se crea la carpeta Letras con sus respectivos archivos
mkdir -p Practica1/Letras
touch Practica1/Letras/a.txt Practica1/Letras/b.txt Practica1/Letras/c.txt

# Dentro de Practica1, se crea la carpeta Integrantes con los nombres del equipo
mkdir -p Practica1/Integrantes
touch Practica1/Integrantes/ChavezSanabriaJoseAdrian.txt
touch Practica1/Integrantes/HernandezGarciaAlejandro.txt
touch Practica1/Integrantes/MartinezPerezJoseAntonio.txt

# Diagrama de la estructura de Practica1
echo "Estructura de la carpeta Practica1:"
if command -v tree &> /dev/null; then
    tree Practica1
else
    echo "El comando 'tree' no está instalado. Utilizando 'ls -R' como alternativa:"
    ls -R Practica1
fi

# Eliminar la carpeta Practica1, incluyendo subcarpetas y archivos
echo "Eliminando la carpeta Practica1 y su contenido..."

# Se comprueba la ruta exacta antes de eliminar
if [ -d "$HOME/Practica1" ]; then
    rm -rf "$HOME/Practica1"
    echo "Carpeta Practica1 eliminada exitosamente."
else
    echo "Error: No se encontró la ruta $HOME/Practica1 para eliminar."
fi

```

---

3. *Explicación de los Bloques Principales del Script de Bash*

- Mensaje de inicio y navegación: Imprime en consola el aviso de ejecución mediante **echo** y utiliza **cd $HOME** para asegurar que el entorno de trabajo sea la raíz del usuario.

- Verificación y creación de Practica1: Un bloque condicional **if [ ! -d "Practica1" ]** evalua si el directorio no existe para proceder a crearlo con **mkdir**.

- Creación de subcarpetas y archivos: Se usa **mkdir -p** para generar las subcarpetas `Letras` e `Integrantes`, y el comando **touch** para generar los archivos de texto vacíos `a.txt`, `b.txt`, `c.txt`, así como los archivos con los nombres de los integrantes del equipo.

- Diagrama de estructura: Evalua si el comando **tree** está instalado. Si lo está, imprime el arbol visual de directorios; si no, ejecuta **ls -R** como alternativa.

- Eliminación segura: Verifica que la ruta `$HOME/Practica1` corresponda exactamente a la carpeta creada antes de ejecutar el comando **rm -rf**, previniendo el borrado accidental de otros directorios.
